```python
import tensorflow as tf
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
goo = yf.download("GOOGL", start="2005-01-01")
goo.head()
```

```python
ms = yf.download("MSFT", start="2005-01-01")
ms.head()
```

```python
ap = yf.download("AAPL", start="2005-01-01")
ap.head()
```

```python
goo_ar = goo['Close']
print(goo_ar)
```

```python
ms_ar = ms['Close']
print(ms_ar)
```

```python
ap_ar = ap['Close']
print(ap_ar)
```

```python
x_data = []
for i in range(4383):
    x_data.append([goo_ar[i-1], ms_ar[i-1]])

y_data = []
for i in range(4383):
    y_data.append([ap_ar[i-1]])

print(x_data)

```

```python
from tensorflow import keras
from tensorflow.keras import layers
```

```python
model = keras.Sequential()
model.add(layers.Embedding(input_dim=1000, output_dim=64))'
'

# The output of GRU will be a 3D tensor of shape (batch_size, timesteps, 256)
model.add(layers.GRU(256, return_sequences=True, dropout=True))

# The output of SimpleRNN will be a 2D tensor of shape (batch_size, 128)
model.add(layers.SimpleRNN(128))

model.add(layers.Dense(10))

model.summary()
```

```python
model.compile(
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer="sgd",
    metrics=["accuracy"],
)


model.fit(np.array(x_data), np.array(y_data), batch_size=2)
```
